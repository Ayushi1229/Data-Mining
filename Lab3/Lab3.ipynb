### 4)  For each quantitative attribute, calculate its average, standard deviation, minimum, mode, range and maximum values.



```python
print('PassengerId: ')
mean=df['PassengerId'].mean()
print(f'\tMean = {mean}')
sd=df['PassengerId'].std()
print(f'\tStandard deviation = {sd}')
min=df['PassengerId'].min()
print(f'\tMinimum = {min}')
max=df['PassengerId'].max()
print(f'\tMaximum = {max}')
mode=df['PassengerId'].mode()[0]
print(f'\tMode = {mode}')
range= max-min
print(f'\tRange = {range}')
```

```python
data=['PassengerId','Survived','Pclass','Age','SibSp','Parch','Fare']
for col in data:
    print(f"{col} :")
    print("\tMean = ",df[col].mean())
    print("\tStandard Deviation = ",df[col].std())
    print("\tMinimum = ",df[col].min())
    print("\tMaximum = ",df[col].max())
    print("\tMode = ",df[col].mode()[0])
    print("\tRange = ",df[col].max()-df[col].min())
```
