### Step 4: Define Apriori Function
This function finds frequent itemsets of size 1, 2, and 3 with minimum support.

```python
from itertools import combinations


def find_frequent_itemsets(data,min_support):
    n = len(data)
    result = []

    for k in [1,2,3]:
        for items in combinations(data.columns,k):
            mask = data[list(items)].all(axis=1)
            support = mask.sum()/n
            # print(f"{frozenset(items)}->{round(support,2)}")
            if support >= min_support:
                result.append((frozenset(items), round(support,2)))

    return result


# find_frequent_itemsets(data,2)
```

### Step 5: Run Apriori
Set `min_support = 0.6` and display the frequent itemsets.

```python
frequent_itemsets = find_frequent_itemsets(data,min_support=0.6)

for itemset, support in frequent_itemsets:
    print(f"{set(itemset)} -> support: {support}")

```

### Step 6 Display as a DataFrame

```python
result_df = pd.DataFrame(frequent_itemsets, columns=['Itemset','Support'])
result_df
```

# Extra : - >  Define Apriori Function without itertools

```python
def find_frequent_itemsets_no_itertools(data, min_support):
    n = len(data)
    result = []

    # For k = 1
    for i in range(len(data.columns)):
        items = (data.columns[i],)
        mask = data[list(items)].all(axis=1)
        support = mask.sum() / n
        if support >= min_support:
            result.append((frozenset(items), round(support, 2)))

    # For k = 2
    for i in range(len(data.columns)):
        for j in range(i + 1, len(data.columns)):
            items = (data.columns[i], data.columns[j])
            mask = data[list(items)].all(axis=1)
            support = mask.sum() / n
            if support >= min_support:
                result.append((frozenset(items), round(support, 2)))

    # For k = 3
    for i in range(len(data.columns)):
        for j in range(i + 1, len(data.columns)):
            for k in range(j + 1, len(data.columns)):
                items = (data.columns[i], data.columns[j], data.columns[k])
                mask = data[list(items)].all(axis=1)
                support = mask.sum() / n
                if support >= min_support:
                    result.append((frozenset(items), round(support, 2)))

    return result
```

```python
frequent_itemsets = find_frequent_itemsets_no_itertools(data, min_support=0.4)

# Display the output
for items, support in frequent_itemsets:
    print(f"{set(items)} -> {support}")
```
