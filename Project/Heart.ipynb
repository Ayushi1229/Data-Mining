# 22. Bar graph of count of targeted variables.

# 23. Line chart on average cholestrol and age.

```python
avg_chol = df.groupby('age')['chol'].mean()

plt.plot(avg_chol.index, avg_chol.values, color='blue', marker='o')
plt.title('Average Cholesterol by Age')
plt.xlabel('Age')
plt.ylabel('Average Cholesterol')
plt.grid(True)
plt.show()
```

# 24. Pie chart on whether patient has disease or not.

```python
import matplotlib.pyplot as plt

labels = ['No Disease', 'Disease']
sizes = df['target'].value_counts().sort_index() 
colors = ['lightcoral', 'lightgreen']

plt.pie(sizes, labels=labels, colors=colors, autopct='%1.1f%%', startangle=90)
plt.title('Heart Disease Distribution')
plt.axis('equal')  # Equal aspect ratio ensures pie is a circle
plt.show()
```

 # 25.Box Plot Grouped by Category 

```python
df.boxplot(column='chol', by='target', grid=True)
plt.title('Cholesterol Distribution by Heart Disease')
plt.suptitle('')
plt.xlabel('Heart Disease (0 = No, 1 = Yes)')
plt.ylabel('Cholesterol Level')
plt.show()
```
