# 21. Pie Chart – Attrition Flag Distribution

```python
labels = df['Attrition_Flag'].value_counts().index
sizes = df['Attrition_Flag'].value_counts().values

plt.pie(sizes, labels=labels, autopct='%1.1f%%', startangle=140)
plt.title('Customer Attrition Distribution')
plt.axis('equal')
plt.show()
```

# 22. Line Plot – Average Transaction Amount vs Age

```python
df_age_grouped = df.groupby('Customer_Age')['Total_Trans_Amt'].mean()

plt.plot(df_age_grouped.index, df_age_grouped.values, color='green', marker='o')
plt.title('Average Transaction Amount by Age')
plt.xlabel('Age')
plt.ylabel('Avg Transaction Amount')
plt.grid(True)
plt.show()
```

# 23. Bar Plot – Count of Customers by Income Category



```python
income_counts = df['Income_Category'].value_counts()

plt.bar(income_counts.index, income_counts.values, color='orange')
plt.title('Customers by Income Category')
plt.xlabel('Income Category')
plt.ylabel('Count')
plt.xticks(rotation=45)
plt.grid(axis='y')
plt.show()
```

# 24. Line Plot – Total Transactions Over Index (Time Simulation)

```python
plt.plot(df.index, df['Total_Trans_Amt'], color='green')
plt.title('Total Transactions Over Time')
plt.xlabel('Index')
plt.ylabel('Transaction Amount')
plt.grid(True)
plt.show()
```

# 25. Histogram – Credit Limit Distribution

```python
plt.hist(df['Credit_Limit'], bins=20, color='salmon', edgecolor='black')
plt.title('Credit Limit Distribution')
plt.xlabel('Credit Limit')
plt.ylabel('Number of Customers')
plt.grid(True)
plt.show()
```
